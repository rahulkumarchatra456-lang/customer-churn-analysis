```python
conn = sqlite3.connect('customer_churn.db')

sql_query = """
    SELECT name
    FROM sqlite_master
    WHERE type='table';
"""

tables = pd.read_sql(sql_query, conn)

# create dataframe for each table
for table_name in tables['name']:
    df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
    globals()[f"df_{table_name}"] = df
    print(f"Created dataframe: df_{table_name}")

conn.close()
```

```python
# Print table names and column names
conn = sqlite3.connect('customer_churn.db')

for table_name in tables['name']:
    print(f"\nTable Name: {table_name}")
    # Get column information
    columns_query = f"PRAGMA table_info({table_name});"
    columns = pd.read_sql(columns_query, conn)
    print("Columns:")
    print(columns['name'].tolist())

# Close connection
conn.close()
```

```python
# 1. Churn Rate 
```

```python
df.columns
```

```python
churn_rate = df['churn_flag'].mean()*100
print("Churn Rate = ", round(churn_rate,2), "%")
```

```python
# 2. Retention Rate 
retention_rate = 100 - churn_rate
print("retention_rate = ", round(retention_rate,2), "%")
```

```python
# 3. Churn by Plan type 
churn_by_plan = (df.groupby('plan_type')['churn_flag'].mean().mul(100).round(2).reset_index(name='churn_rate_pct'))
print("Churn by plan = ", churn_by_plan)      
```

```python
# 4. ARPU - Avg Revenue per user 
arpu = df['monthly_charges'].mean()
print('ARPU = ',round(arpu,2))
```

```python
# homework - calculate customer age 
```

```python
# 5. Avg Customer Tenure
# count of days users has used our service : cancellation date else current rate 
today = pd.Timestamp.today()

df['tenure_days'] = np.where(
           df['cancellation_date'].notna(),
    
    (df['cancellation_date'] - df['subscription_start_date']).dt.days,
    
    (today - df['subscription_start_date']).dt.days
)

avg_tenure = df['tenure_days'].mean()
print("Avg Tenure (Days) = ", round(avg_tenure),0)
```

```python
# 6. Revenue at risk - revenue last from churned users 
revenue_at_risk = df.loc[df['churn_flag']==1, 'monthly_charges'].sum()
print("Revenue at Risk (Rs 'K') =", revenue_at_risk)
```

```python
# 7. Escalation Rate
escalation_rate = (df['escalations']=='Y').mean()*100
print("Escalation Rate = ", round(escalation_rate, 2), "%")
```

```python
# 8. Avg Complaint Per User
avg_complaints = df['complaint_count'].sum() / df['customerid'].nunique()
print("Avg Complaints Per User = ", round(avg_complaints,2))
```

```python

# 9. Correlation: Escalation vs Churn

df['escalation'] = np.where(df['escalations'] == 'Y', 1, 0)
corr_df = df[['escalation', 'churn_flag']].dropna()

correlation = corr_df['escalation'].corr(corr_df['churn_flag'])
print("Correlation between escalation vs churn is =", round(correlation, 2))

```

```python

# 10. Create a column using existing column - Churn Risk  

conditions = [
    (df['churn_score'] < 50),
    (df['churn_score'] >= 50) & (df['churn_score'] < 70),
    (df['churn_score'] >= 70)
]

choices = ['low', 'med', 'high']

df['churn_risk'] = np.select(
    conditions,
    choices,
    default='unknown'
)
```

```python
# 4.1 Monthly Churn Trend (Time Series KPI)

df_visual['cancellation_month'] = df_visual['cancellation_date'].dt.to_period('M')

churn_trend = df_visual[df_visual['churn_flag'] == 1].groupby('cancellation_month').size()
plt.figure(figsize=(8,3))
plt.plot(churn_trend.index.astype(str), churn_trend.values, color='green',marker='o',linestyle='dashed', linewidth=2, markersize=12)

plt.title('Monthly Churn Trend')
plt.xlabel('Month')
plt.ylabel('Churned Customers')
plt.show()
```

```python
# 4.2 Churn by Plan type 
churn_plan = df_visual.groupby('plan_type')['churn_flag'].mean()

colors = ['grey', 'gold', 'teal']

plt.figure(figsize=(7,4))

plt.bar(churn_plan.index, churn_plan.values, color = colors)

plt.title('Churn Rate by Plan Type')
plt.xlabel('Plan Type')
plt.ylabel('Churned Rate')
plt.show()
```

```python
# 4.3 Churn by States
churn_plan = df_visual.groupby('state')['churn_flag'].mean()

#colors = ['grey', 'gold', 'teal']
colors = plt.cm.Set2(np.linspace(0, 1, len(churn_plan)))

plt.figure(figsize=(12,4))

plt.bar(churn_plan.index, churn_plan.values, color = colors)

plt.title('Churn Rate by State')
plt.xlabel('State')
plt.ylabel('Churned Rate')
plt.show()
```

```python
# 4.4 Churn by Contract Type 
churn_plan = df_visual.groupby('contract_type')['churn_flag'].mean()

#colors = ['grey', 'gold', 'teal']
colors = plt.cm.Set2(np.linspace(0, 1, len(churn_plan)))

plt.figure(figsize=(4,4))

plt.bar(churn_plan.index, churn_plan.values, color = colors)

plt.title('Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churned Rate')
plt.show()
```

```python
# 4.5 Churn by Cancellation Reason
churn_plan = df_visual.groupby('cancellation_reason')['churn_flag'].mean()

#colors = ['grey', 'gold', 'teal']
colors = plt.cm.Set2(np.linspace(0, 1, len(churn_plan)))

plt.figure(figsize=(12,4))

plt.bar(churn_plan.index, churn_plan.values, color = colors)

plt.title('Top Reasons for Customer Churn')
plt.xlabel('Cancellation Reason')
plt.ylabel('Churned Customers')
plt.show()
```

```python
# 4.6 Churn by Customer Satisfaction
churn_plan = df_visual.groupby('csat_score')['churn_flag'].mean()

#colors = ['grey', 'gold', 'teal']
colors = plt.cm.Set2(np.linspace(0, 1, len(churn_plan)))

plt.figure(figsize=(7,4))

plt.bar(churn_plan.index, churn_plan.values, color = colors)

plt.title('Churn Rate by Customer Satisfaction')
plt.xlabel('CSAT Score')
plt.ylabel('Churned Rate')
plt.show()
```
